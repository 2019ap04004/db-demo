# Extract Data from the refunds sql table
1. create bronze schema in hive metastore
2. create external table

```sql
create schema if not exists hive_metastore.bronze;
```

```sql
-- Since we are using Databricks serverless compute below code won't work as JDBC is not supported by databricks serverless compute
create table if not exists hive_metastore.bronze.refunds
using JDBC
OPTIONS (
  url 'jdbc:sqlserver://gizmoboxoy.database.windows.net:1433;database=gizmobox-db',
  dbtable 'refunds',
  user 'gizmoadmin',
  password '<Password>'
)
```

```sql
%python
# PySpark or Databricks notebook example

jdbc_url = "jdbc:sqlserver://gizmoboxoy.database.windows.net:1433;database=gizmobox-db"
connection_properties = {
    "user": "gizmoadmin",
    "password": "<Password>",
    "driver": "com.microsoft.sqlserver.jdbc.SQLServerDriver"
}

# Read from SQL Server
refunds_df = spark.read.jdbc(url=jdbc_url, table="refunds", properties=connection_properties)

# Write to Delta (in Hive metastore bronze schema)
refunds_df.write.format("delta").mode("overwrite").saveAsTable("hive_metastore.bronze.refunds")
```
