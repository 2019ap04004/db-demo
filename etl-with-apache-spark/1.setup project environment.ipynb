# Configure access to cloud storage via unity catalog

## create external location

```sql
create external location if not exists dea_ext_dl_gizmobox
url 'abfss://gizmobox@deaoyextdl1.dfs.core.windows.net/'
with (storage CREDENTIAL `dea-ext-ac`)
comment 'DEA external' 
```

## access the container gizmobox

```sql
%fs ls 'abfss://gizmobox@deaoyextdl1.dfs.core.windows.net/landing'
```

```sql
show catalogs;
```

## create a catalog gizmobox

```sql
create catalog if not exists gizmobox
managed location 'abfss://gizmobox@deaoyextdl1.dfs.core.windows.net/'
comment 'gizmobox catalog'
```

```sql
show catalogs;
```

### create 4 schemas
1. landing
2. bronze
3. silver
4. gold

```sql
select current_catalog();
```

```sql
use catalog gizmobox;
```

```sql
select current_catalog();
```

```sql
create schema if not exists landing
comment 'we will have landing data here'
managed location 'abfss://gizmobox@deaoyextdl1.dfs.core.windows.net/landing';
create schema if not exists bronze
comment 'we will have bronze data here'
managed location 'abfss://gizmobox@deaoyextdl1.dfs.core.windows.net/bronze';
create schema if not exists silver
comment 'we will have silver data here'
managed location 'abfss://gizmobox@deaoyextdl1.dfs.core.windows.net/silver';
create schema if not exists gold
comment 'we will have gold data here'
managed location 'abfss://gizmobox@deaoyextdl1.dfs.core.windows.net/gold';
```

```sql
show schemas;
```

use Volume

```sql
use catalog gizmobox;
use schema landing;

create external volume if not exists operational_data
location 'abfss://gizmobox@deaoyextdl1.dfs.core.windows.net/landing/operational_data'
comment 'operational data volume';

```

```sql
%fs ls /Volumes/gizmobox/landing/operational_data
```
