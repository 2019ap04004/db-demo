# Extract Data from order JSON File
1. Query order file using JSON Format.
2. Query order file using TEXT Format.
3. Query orders view in bronze schema.


> **_NOTE:_** syntax for querying a file is
> ```
>  select * from <FILE_TYPE>.`<FILE_PATH>`
> ```

### 1. Query order file using JSON Format.

````sql
select * from text.`/Volumes/gizmobox/landing/operational_data/orders`;
````

### 2. Query order file using TEXT Format.

````sql
select * from text.`/Volumes/gizmobox/landing/operational_data/orders`;
````

### 3. Create a view in bronze schema

````sql
create or replace view gizmobox.bronze.v_orders as
select * from text.`/Volumes/gizmobox/landing/operational_data/orders`;
````

````sql
select * from gizmobox.bronze.v_orders
````
