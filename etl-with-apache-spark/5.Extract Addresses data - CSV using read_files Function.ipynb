#Extract Data from the Address Files
1. Demonstrate Limitations of CSV file format in select statement
2. Use read_files function to overcome the limitations
3. Create addresses view in the bronze layer.

### 1. Demonstrate Limitations of CSV file format in select statement

```sql
select * from csv.`/Volumes/gizmobox/landing/operational_data/addresses/`
```

### 2. Use read_files function to overcome the limitations

```sql
select * from read_files('/Volumes/gizmobox/landing/operational_data/addresses', format=>'csv', delimiter=>'\t', header=>True)
```

### 3. Create addresses view in the bronze layer.

```sql
create or replace view gizmobox.bronze.v_addresses as select * from read_files('/Volumes/gizmobox/landing/operational_data/addresses/', format=>'csv', delimiter=>'\t', header=>True);
```
